1.	Data Collection: The company has a dataset containing the following fields:
○	EmployeeID: Unique identifier for each employee.
○	Age: Age of the employee.
○	Gender: Gender of the employee.
○	Job Title: Employee’s role in the company.
○	Salary: Current salary of the employee.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('employees_data.csv')
df
```

2.	Data Preprocessing: Before proceeding with analysis, the data is cleaned, with missing or erroneous values handled appropriately (e.g., filling in missing ages, handling gender misclassifications). The data is then loaded into a Pandas DataFrame for further analysis.

```python
# missing values
print(df.isnull().sum())

# remove duplicates
df.drop_duplicates(inplace=True)

# Data Types
print(df.dtypes) 
```

3.	Data Analysis:
○	Do profiling on data to generate an in-depth summary of the dataset. This report includes key metrics such as missing values, variable distributions, and correlations between variables.

○	The profiling report highlights trends in the data, such as the relationship between salary and job title, the distribution of salaries by gender, and possible outliers in the salary data.

```python

avg_salary = df.groupby('JobTitle')['Salary'].mean()
print(avg_salary)
```

```python
avg_salary= df.groupby('JobTitle')['Salary'].mean()
print(avg_salary)
```

```python
#  Analyze Salary Distribution Across Job Titles

avg_salary.plot(kind="bar")
plt.title("Average Salary by Job Title")
plt.xlabel("Job Title")
plt.ylabel("Average Salary")
plt.xticks(rotation=45)
plt.show()
```

```python
# Analyze Employee Demographics

plt.hist(df["Age"], bins=10)
plt.title("Employee Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.show()
```

```python
# Gender Distribution

gender_count = df["Gender"].value_counts()

print(gender_count)

gender_count.plot(kind="bar")
plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
# Analyze Salary by Gender

salary_gender = df.groupby("Gender")["Salary"].mean()

print(salary_gender)
```

```python
salary_gender.plot(kind="bar")
plt.title("Average Salary by Gender")
plt.ylabel("Salary")
plt.show()
```

```python
# Analyze Salary by Location

location_salary = df.groupby("Location")["Salary"].mean()

print(location_salary)
```

```python
location_salary.plot(kind="bar")
plt.title("Average Salary by Location")
plt.ylabel("Salary")
plt.show()
```

```python
# Correlation Analysis

correlation = df[["Age", "Salary"]].corr()

print(correlation)
```
