#### 1. Basic Questions:-
1. What is histogram?
>>A histogram is a graphical tool used in statistics to visualize the frequency distribution of continuous numerical data. It groups data into contiguous numerical ranges called "bins" or "classes" and uses the height or area of each vertical bar to represent the number of data points that fall within that specific range

2. What is a bin?  >>  (e.g., 1–10, 10–20, or 0.5 to 1.5).
>>A bin in a histogram is a range of continuous values used to group raw numerical data

3. What does frequency mean?
>> Frequency refers to the number of times an event or occurrence repeats within a specific period.

4. Why are histograms useful?
    1. Visualizing Data Distribution
    2. Identifying Outliers
    3. Spotting Patterns & Trends
    4. Assessing Process Capability

#### 2. Numerical Problems:-
---
Given the data: 12, 15, 18, 20, 22, 25, 27, 30
---
1. Create bins of size 10.

2. Find the frequency in each bin

--- 
Given the scores: 45, 50, 55, 60, 65, 70, 75, 80
---
1. Create a histogram.
2. Identify the most common score range.

```python
# 1. Create bins of size 10.

# Using NumPy

import numpy as np

data = [12, 15, 18, 20, 22, 25, 27, 30]

# Create bins of size 10
bins = np.arange(0, max(data) + 10, 10)

print(bins)
```

```python
# Using Pandas to categorize data into these bins

import pandas as pd
import numpy as np

data = [12, 15, 18, 20, 22, 25, 27, 30]

bins = np.arange(0, max(data) + 10, 10)

categories = pd.cut(data, bins)    #  pd.cut()  -  checks each value and decides which interval it belongs to.

print(categories)
```

```python
# 2. Find the frequency in each bin

import pandas as pd
import numpy as np

data = [12, 15, 18, 20, 22, 25, 27, 30]

bins = np.arange(0, max(data) + 10, 10)

categories = pd.cut(data, bins=bins)

frequency = categories.value_counts().sort_index()

print(frequency)
```

```python
# --- 
# Given the scores: 45, 50, 55, 60, 65, 70, 75, 80
# ---
# 1. Create a histogram.

import matplotlib.pyplot as plt

# Data
scores = [45, 50, 55, 60, 65, 70, 75, 80]

# Create histogram
plt.hist(scores, bins=4, edgecolor='black')

# Labels
plt.xlabel("Scores")
plt.ylabel("Frequency")
plt.title("Histogram of Scores")

# Show plot
plt.show()
```

```python
# 2. Identify the most common score range.

import numpy as np

scores = [45, 50, 55, 60, 65, 70, 75, 80]

bins = np.arange(40, 90, 10)

freq, edges = np.histogram(scores, bins=bins)

print("Most common range:", 
      (int(edges[np.argmax(freq)]), int(edges[np.argmax(freq)+1])))
```

>>> A bar chart is used for categorical data (like fruits, cities, subjects) and compares separate categories.

2. What happens if the bin size is too large?
>>> You lose important information.

3. What happens if the bin size is too small?
>>> You lose clarity of the trend.

4. How can a histogram help identify outliers?
>>> Outliers appear as separate small bars away from the bulk of data.

#### 1. Basic Questions:-
1. What is a Box Plot?
>>> A Box Plot (Box and Whisker Plot) is a graphical way to show how numerical data is distributed using five key summary values.
>>> It helps us understand:
    spread of data,
    central value,
    outliers
2. What are the five values used in a Box Plot?
>>> A box plot is built using the five-number summary:
    Minimum – smallest value
    First Quartile (Q1) – 25% data point
    Median (Q2) – middle value (50%)
    Third Quartile (Q3) – 75% data point
    Maximum – largest value

3. What does the median represent?
>>> The median is the middle value of the data when arranged in order.

4. What is the purpose of whiskers?
>>> Whiskers are the lines extending from the box in a box plot.

#### 2. Numerical Problems:-

#### 3. Conceptual Questions:- 
1. Why is IQR useful?
>>>> IQR (Interquartile Range) is useful because it shows the spread of the middle 50% of your data, and it is not affected by extreme values (outliers).IQR=Q3−Q1

2. Why are Box Plots good for detecting outliers?
>>> Box plots are good for detecting outliers because they clearly separate the “normal” data from extreme values using a standard rule based on the IQR (Interquartile Range).

3. What information does the box represent?
>>> The box in a box plot represents the middle 50% of the data, also called the interquartile range (IQR).

4. What does a long whisker indicate?
>>> A long whisker in a box plot indicates that the data has a wide spread outside the middle 50% in that direction.

```python
# 1. Create a Bar Chart using:
#     - Product A = 50
#     - Product B = 80
#     - Product C = 30
#     - Product D = 100

import matplotlib.pyplot as plt

# Data
products = ["Product A", "Product B", "Product C", "Product D"]
values = [50, 80, 30, 100]

# Bar chart
plt.bar(products, values)

# Labels and title
plt.xlabel("Products")
plt.ylabel("Values")
plt.title("Product Performance")

# Show plot
plt.show()
```

```python
# 2. Create a Line Chart using:
#     - Jan 200
#     - Feb = 250
#     - Mar 280
#     - Apr = 320
#     - May = 350

import matplotlib.pyplot as plt

# Data
months = ["Jan", "Feb", "Mar", "Apr", "May"]
values = [200, 250, 280, 320, 350]

# Line chart
plt.plot(months, values, marker='+')

# Labels and title
plt.xlabel("Months")
plt.ylabel("Values")
plt.title("Monthly Trend")

# Show plot
plt.show()
```

```python
# 3. Identify the highest and lowest values.

# Data
months = ["Jan", "Feb", "Mar", "Apr", "May"]
values = [200, 250, 280, 320, 350]

# Find highest value
max_value = max(values)
max_month = months[values.index(max_value)]

# Find lowest value
min_value = min(values)
min_month = months[values.index(min_value)]

# Output
print("Highest value:", max_value, "in", max_month)
print("Lowest value:", min_value, "in", min_month)
```
